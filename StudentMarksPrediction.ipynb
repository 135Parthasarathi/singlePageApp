```python
import numpy as np
import pandas as pd
```

```python
rng = np.random.default_rng()
```

```python
attendence = rng.integers(0,101,size=(64))
print(attendence)
```

```python
studyhours = rng.integers(0,11,size=(64))
print(studyhours)
```

```python
dataset=np.vstack((attendence,studyhours))
print(dataset)
```

```python
dataset=dataset.T
print(dataset)
```

```python
w1=0.4
w2=5
bias=8
```

```python
marks=(w1*attendence+w2*studyhours)+bias
print(marks)
```

```python
dataset = np.hstack((dataset,np.ones((64,1))))
print(dataset)
```

```python
w=np.linalg.inv(dataset.T @ dataset)@dataset.T@marks
print(w)
```

```python
data = np.vstack((attendence,studyhours,marks)).T
print(data)
```
